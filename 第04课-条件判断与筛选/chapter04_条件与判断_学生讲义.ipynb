## 1. 比较运算与布尔值

### 理论知识

**比较运算用来判断两个值的关系，结果只有两种：`True` 或 `False`。** 这种结果的类型是 `bool`（布尔）。

| 运算 | 含义 |
| --- | --- |
| `==` | 相等 |
| `!=` | 不相等 |
| `>` `>=` | 大于、大于或等于 |
| `<` `<=` | 小于、小于或等于 |

比较不会修改原来的值；它只产生一个判断结果。赋值 `=` 与比较 `==` 不是同一件事：前者给名字，后者问是否相等。

两个运算元的类型会影响比较含义。数字按数值大小比较；字符串按文本比较，不是按数值大小。看起来像数字的字段，若仍是 `str`，就还不是数值比较。

### 案例：一条记录上的判断结果


````python
platform_text = "TRUE"
score_text = "8.8"
print(platform_text == "TRUE")
print(platform_text == True)
print(type(platform_text == "TRUE"))
print(score_text >= "8.5")
print(float(score_text) >= 8.5)
````

### 讲解

第一个比较为 `True`，因为两边都是文本 `"TRUE"`。第二个为 `False`：左边是字符串，右边是布尔值，二者不相等。`type(...)` 显示比较结果的类型是 `bool`。

`"8.8" >= "8.5"` 是字符串比较，碰巧也得到 `True`，但不能当成评分规则。`float(score_text) >= 8.5` 才是数值比较。后文凡是按评分高低筛选，都先 `float`。

### 易错点与练习

把 `==` 写成 `=` 不是“比错了”，而是改成了赋值，常常直接语法错误，或把条件位置变成意外赋值。排查时先看是不是少写了一个等号。

## 2. if 语句：条件成立才执行

### 理论知识与语法

**`if` 根据一个布尔值决定要不要执行某段代码。** 基本形式：

```python
if condition:
    statement_1
    statement_2
statement_after
```

- `if` 和冒号是语法的一部分；`condition` 写比较或已经得到的布尔值。
- 冒号后面缩进的语句组成 `if` 体；本课统一四个空格。
- `condition` 为 `True` 时执行缩进语句；为 `False` 时跳过它们。
- `statement_after` 没有缩进，无论条件真假都会执行。

`if` 本身不会自动遍历表格。它一次只判断当前这个条件。要检查每一行，需要和 `for` 一起用：`for` 负责取出下一行，`if` 负责这一行要不要处理。

### 案例：只在平台字段匹配时打印


````python
one_game = ["人格解体", "8.8", "角色扮演", "TRUE"]
if one_game[3] == "TRUE":
    print("Keep:", one_game[0])
print("Finished one record")
````

### 讲解

`one_game[3] == "TRUE"` 为 `True`，因此打印名称。最后一句没有缩进，判断结束后一定会执行。若把平台改成 `"FALSE"`，中间的 `print` 不会运行，但结束提示仍会打印。

下面用五条摘录记录跟踪：哪些行会进入 `if` 体，哪些行只经过循环、不打印名称。简化表的字段为 title、score、game_type、isAndroid，下标依次是 0、1、2、3。这与完整七列表头不同，后文读取文件时会改回完整下标。

## 4. 在循环中筛选：留下安卓记录

### 理论知识与语法

筛选一张表，标准写法是：**遍历每一行，用 `if` 判断是否留下，把留下的行放进新列表。**

```python
selected = []
for row in body:
    if condition:
        selected.append(row)
```

`append` 把当前行加到 `selected` 末尾。它不是新的数据类型，只是列表“在末尾增加一个元素”的方法。本课用它收集筛选结果，不展开其它列表方法。

- `selected` 必须在循环前建成空列表，循环内不要重新写成 `[]`。
- `condition` 为 `False` 的行不会 `append`，也就不会进入后续平均分。
- `len(selected)` 是留下的记录数，作为子集平均分的分母候选。

### 案例：收集 isAndroid 为文本 TRUE 的记录


````python
android_rows = []
for row in body:
    if row[6] == "TRUE":
        android_rows.append(row)
print("Android count:", len(android_rows))
print("First android title:", android_rows[0][1])
print("Original record count:", len(body))
````

### 讲解

全表 266 条记录里，平台字段等于 `"TRUE"` 的被放进 `android_rows`。`body` 本身没有被删改；筛选得到的是另一份列表。打印 `len(android_rows)` 与 `len(body)`，可以看见子集和全表不是同一个范围。

只想看前几条匹配结果时，仍先按全表规则筛选，再对结果切片，不要把 `body[:5]` 误当成“全表安卓”。

### 易错点与练习

`if row[6] = "TRUE"` 少写等号。`if row == "TRUE"` 拿整行去和字符串比，几乎总是 `False`。循环内写 `android_rows = []` 会每轮清空已收集的行。

## 5. 文本 "TRUE" 与布尔 True

### 理论知识

CSV 读进来的字段默认是字符串。本文件的 `isAndroid` 取值是文本 `"TRUE"` 和 `"FALSE"`，**不是**布尔值 `True` 和 `False`。

判断平台时应写：

```python
row[6] == "TRUE"
```

不要写 `row[6] == True`，也不要写 `if row[6]:`。非空字符串在 `if` 里会被当成真，那样 `"FALSE"` 也会被留下，因为 `"FALSE"` 仍是非空文本。

`type(row[6])` 应显示 `str`。是否安卓，以文件里的真实文本为准。

### 案例：同一条记录的三种判断


````python
platform_field = body[0][6]
print("Raw value:", platform_field, type(platform_field))
print("Equal to text TRUE:", platform_field == "TRUE")
print("Equal to boolean True:", platform_field == True)
if platform_field:
    print("Non-empty text is treated as true in if")
false_text = "FALSE"
print("FALSE text in if:", bool(false_text))
print("FALSE text == TRUE:", false_text == "TRUE")
````

### 讲解与易错点

第一行平台文本是 `'TRUE'`，与字符串比较为真，与布尔 `True` 比较为假。`if platform_field:` 只问“这串文本空不空”，因此 `"TRUE"` 和 `"FALSE"` 都会走进 `if` 体。筛选安卓必须用 `== "TRUE"`。

`bool("FALSE")` 为 `True` 常让人意外。本课不把 `bool(...)` 当作平台转换函数。

## 6. != 筛选非安卓记录

### 理论知识与语法

`!=` 表示不相等。找非安卓记录时，与文件中的安卓标记相反即可：

```python
row[6] != "TRUE"
```

在本文件中，平台字段只有 `"TRUE"` 和 `"FALSE"`，因此 `row[6] != "TRUE"` 与 `row[6] == "FALSE"` 选出同一批记录。书写时仍以“不是安卓标记”这个规则为准，先核对表里实际出现过哪些取值。

若分类完整且互不重叠，则：

```text
安卓记录数 + 非安卓记录数 = 数据记录数
```

这个等式用来检查有没有漏行，不是新的平均分公式。

### 案例：对照安卓与非安卓数量


````python
not_android_rows = []
for row in body:
    if row[6] != "TRUE":
        not_android_rows.append(row)
print("Android:", len(android_rows))
print("Not android:", len(not_android_rows))
print("Sum:", len(android_rows) + len(not_android_rows))
print("Body:", len(body))
````

### 讲解

两份列表数量相加应等于 `len(body)`。不相等时，优先检查条件是否写成了 `== True`、是否把表头算进去，或是否在循环里清空了列表。

`sample` 里非安卓是后两条：另一个伊甸与偶像梦幻祭2。完整文件不要用手算 266 条来验收，用程序计数并核对守恒即可。

### 易错点与练习

`row[6] != True` 几乎对每一行都为真，因为字段是字符串。守恒失败时不要先改平均分公式，先打印两个 `len`。

## 7. 子集平均分：只累计被留下的评分

### 理论知识与公式

筛选后的平均分，分子和分母必须来自**同一批被留下的记录**：

```text
子集平均分 = 留下记录的评分总和 / 留下记录的条数
```

循环里同时完成两件事：判断是否留下，以及（若留下）把评分转成浮点数后累加。计数器也只在条件成立时加 1。

```python
total = 0.0
count = 0
for row in selected_rows:
    score = float(row[score_index])
    total = total + score
    count = count + 1
average = total / count
```

也可以直接在全表循环里写 `if`，不必先建 `selected_rows`；两种写法的口径必须一致。`count` 为 0 时不能相除。本课演示中的子集均非空。

### 案例：角色扮演记录的平均分


````python
rpg_total = 0.0
rpg_count = 0
for row in body:
    if row[3] == "角色扮演":
        rpg_total = rpg_total + float(row[2])
        rpg_count = rpg_count + 1
rpg_average = rpg_total / rpg_count
print("RPG count:", rpg_count)
print("RPG total:", rpg_total)
print("RPG mean:", rpg_average)
````

### 代码讲解

只有 `game_type` 为 `"角色扮演"` 的行参与累加。分母是 `rpg_count`，不是 `len(body)`，也不是字段个数 7。全表平均分会掩盖“只看角色扮演”的结果，这正是本课要做筛选的原因。浮点输出可能出现末位尾差，不改变这些记录被计入的口径。

## 8. and：同时满足

### 理论知识与语法

`and` 连接两个条件，**左右都为 `True` 时，整个表达式才为 `True`**。

```python
condition_a and condition_b
```

| 左 | 右 | `and` |
| --- | --- | --- |
| False | False | False |
| False | True | False |
| True | False | False |
| True | True | True |

它不是把两个筛选结果的行数相加。它要求同一条记录同时符合两条规则，例如“类型是角色扮演，并且平台是安卓”。

### 案例：角色扮演且安卓

## 9. or：满足其一

### 理论知识与语法

`or` 连接两个条件，**左右至少有一个为 `True` 时，整个表达式为 `True`**。两个都为真也会留下，不是“只能选一边”。

```python
condition_a or condition_b
```

| 左 | 右 | `or` |
| --- | --- | --- |
| False | False | False |
| False | True | True |
| True | False | True |
| True | True | True |

日常口语里的“或者”有时被理解成互斥。Python 的 `or` 不是互斥：角色扮演并且同时也能被其它规则说中的记录，仍然只是一条记录、会被留下一次。

源实验材料里若把 `or` 说成“同时满足”，以课堂规则为准：**同时满足是 `and`，满足其一是 `or`。**

### 案例：角色扮演或策略


````python
print("Sample traces:")
for row in sample:
    is_rpg = row[2] == "角色扮演"
    is_strategy = row[2] == "策略"
    print(row[0], is_rpg, is_strategy, is_rpg or is_strategy)

rpg_or_strategy = []
for row in body:
    if row[3] == "角色扮演" or row[3] == "策略":
        rpg_or_strategy.append(row)
print("Full-file RPG or strategy:", len(rpg_or_strategy))
````

### 讲解

`sample` 中偶像梦幻祭2 的类型是音乐，两个条件都为假，被排除；其余四条至少满足一侧。人格解体只满足左侧，明日方舟只满足右侧，它们都会留下。这就是“满足其一”，不是“两个条件要一起成立”。

完整文件里类型只有一个取值，一条记录不会既是角色扮演又是策略；即便如此，`or` 的定义仍然包含“两侧都真”的那一行。

## 10. 组合条件与括号

### 理论知识

实际筛选常常是两类规则叠在一起，例如：**类型为角色扮演或策略，并且不是安卓。**

写成：

```python
row[6] != "TRUE" and (row[3] == "角色扮演" or row[3] == "策略")
```

括号是必要的。在 Python 中 `and` 的结合优先于 `or`：没有括号时，`A and B or C` 相当于 `(A and B) or C`，口径就变了。先用中文说出规则，再给“且 / 或”两边加上完整比较，最后用括号把“或”包在一起。

## 12. if / elif / else：分档与计数守恒

### 理论知识与语法

一条记录在同一套分档规则里只应落入一档。`elif` 表示“上面都不成立时，再问下一个条件”；`else` 承接剩余情况。

```python
if score >= 8.5:
    high = high + 1
elif score >= 7.0:
    mid = mid + 1
else:
    low = low + 1
```

顺序有含义：先问高分，再问中分。若先写 `>= 7.0`，高分记录也会先进中档，后面的 `>= 8.5` 再也轮不到。三个分支互斥，因此：

```text
high + mid + low = 参与分档的记录数
```

对全表分档时，右侧是 `len(body)`。不要用三个独立的 `if`（不用 `elif`）去加三个计数，同一条高分记录可能被多个条件重复统计。

### 案例：sample 五条记录分档


````python
high = 0
mid = 0
low = 0
for row in sample:
    score = float(row[1])
    if score >= 8.5:
        high = high + 1
    elif score >= 7.0:
        mid = mid + 1
    else:
        low = low + 1
print("High / mid / low:", high, mid, low)
print("Sum:", high + mid + low)
print("Sample size:", len(sample))
````

### 讲解

8.8 与 9 记入高分，两条 8.2 记入中分，6.5 记入低分。2 + 2 + 1 = 5，与 `len(sample)` 一致。全表分档把 `row[1]` 换成 `row[2]`，并把循环对象换成 `body`；具体三个计数留给综合练习独立完成。

### 易错点与练习

三个都写成 `if` 会破坏互斥。漏掉 `else` 时，低于 7.0 的记录不进入任何档，守恒失败。阈值写反（先 7.0 后 8.5）时，高分会被中档吃掉。
