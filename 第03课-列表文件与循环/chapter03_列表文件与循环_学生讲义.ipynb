## 1. 列表的定义与创建

### 理论知识

**列表（list）是一种有序、可修改的序列。** 它把多个元素保存在一个对象中，用于表示一组相关数据。

- **有序**：元素保留放入时的位置，可以按位置访问；有序不等于自动按大小排序。
- **可修改**：创建后可以替换其中的元素，不必重新创建整条记录。
- **元素类型不必相同**：一个列表可以同时含字符串和数值；每个元素仍有自己的类型。
- **允许重复**：相同的值可以出现在多个位置，列表不会自动去重。

基本语法：

```python
items = [element1, element2, element3]
empty_items = []
```

这里element1等表示准备放入的值或已有变量，不是必须使用的名称。方括号创建列表，逗号分隔元素；引号决定某个元素是不是字符串。

### 案例：一个列表对象与它的元素

````python
items = ["明日方舟", 8.2, "策略"]
empty_items = []
print(items)
print(type(items))
print(empty_items)
````

### 讲解

items整体的类型是list，其中`"明日方舟"`和`"策略"`是字符串，8.2是浮点数。empty_items也是列表，只是还没有元素。

原CSV中的所有字段都是文本。摘取同一条记录的title、score、game_type时，可以先保留文本类型：

````python
one_game = ["明日方舟", "8.2", "策略"]
print(one_game)
print(type(one_game))
````

## 2. 列表索引：按位置访问元素

### 理论知识与语法

**索引（index）是元素的位置编号。** 使用`items[index]`取出一个元素，原列表不会因为取值而改变。

- 非负索引从左往右：0表示第一项，1表示第二项。
- 负索引从右往左：-1表示最后一项，-2表示倒数第二项。
- 长度为n的非空列表，合法索引范围是`0`到`n-1`，或`-n`到`-1`。
- 索引必须是整数形式；`items[1]`与`items["1"]`不是同一种访问。
- 访问结果的类型由被选元素决定，索引操作不会自动转换类型。

以原记录摘取的三字段列表为例：

| 元素 | 明日方舟 | 8.2（文本） | 策略 |
| --- | --- | --- | --- |
| 非负索引 | 0 | 1 | 2 |
| 负索引 | -3 | -2 | -1 |

### 案例：位置与取值

````python
one_game = ["明日方舟", "8.2", "策略"]
print(one_game[0])
print(one_game[-1])
print(one_game[1], type(one_game[1]))
````

第一个print取名称，第二个取最后的类型字段，第三个取评分文本并检查类型。评分看起来是数字，但类型仍为str。

## 3. 修改列表元素：元素赋值与变量重新赋值

### 理论知识与语法

列表的**可修改性**指可以改变已有位置的元素。基本语法是：

```python
items[index] = new_value
```

先确定items所指的列表，再找到index位置，用new_value替换该位置的旧元素。其他位置保持不变，单元素替换不改变列表长度；被替换元素的类型也可以改变。

这与`items = new_list`不同：后者让变量名指向另一个列表。`items[index]`作为赋值目标时表示“修改这个位置”；仅出现在print中时表示“取出这个位置的值”。

### 案例：把评分文本转换后放回原位置

````python
editable_record = ["明日方舟", "8.2", "策略"]
print(editable_record[1], type(editable_record[1]))
editable_record[1] = float(editable_record[1])
print(editable_record[1], type(editable_record[1]))
print(editable_record)
````

### 代码讲解与数据案例

右侧`float(editable_record[1])`先把文本转成浮点数8.2，再赋回同一位置。因此评分的数值含义没变，但类型从str变为float；名称和类型字段保持不变。

原始CSV的内容没有被改写。Python中的内存对象与硬盘文件是两个层次：修改列表不是保存文件。下面重新用原记录建立变量，可以再次得到评分文本。

````python
original_record = ["明日方舟", "8.2", "策略"]
print(original_record[1], type(original_record[1]))
````

### 易错点与练习

`items[3] = value`不能给三个元素的列表添加第四项，会越界。单个元素赋值只能替换已存在的位置，本节不使用它扩展列表。

## 4. len：长度不是最后一个索引

### 理论知识与语法

`len(object)`是内置函数，返回对象包含的元素数量，结果为整数。对列表，它数最外层元素；对字符串，它数该字符串中的字符。本节主要用它区分记录的字段数和表格的记录数。

```python
number_of_items = len(items)
```

非空列表有n项，最后一个非负索引是n-1。空列表长度为0，没有可以访问的第一个或最后一个元素。因此“长度是多少”与“最后位置在哪里”是两个问题。

### 代码与数据案例

````python
one_game = ["明日方舟", "8.2", "策略"]
print("Field count:", len(one_game))
print("Last nonnegative index:", len(one_game) - 1)
print("Last element:", one_game[len(one_game) - 1])
print("Empty list length:", len([]))
````

### 代码讲解与易错点

三个字段的数量是3，最后非负索引为2，所取元素为策略。len([])为0，但不能因此访问空列表的索引0。

不要把`len(one_game)`当成“这张表有几个游戏”：one_game是一条记录，len数的是它的字段。函数返回的数字含义取决于传入的对象。

## 5. 嵌套列表与二维索引

### 理论知识与语法

当列表的某个元素本身也是列表时，就形成嵌套结构。若外层每个元素都代表一条记录，便可以用“列表的列表”表示表格。

```python
table = [row1, row2, row3]
value = table[row_index][column_index]
```

`table[row_index]`先返回一个内层列表，再对该列表用`[column_index]`选字段。执行过程从左往右，不是同时在一个平面上寻找两个数字。

- table是整张表；table[i]是一条记录；table[i][j]是单个字段。
- len(table)数行；len(table[i])数第i行的字段。
- Python允许内层列表长度不同，不会自动检查它是否为规则表格。使用固定列之前，应确认输入有一致的字段结构。

### 代码：把一次连续访问拆成两步

````python
small_table = [
    ["人格解体", "8.8", "角色扮演"],
    ["一念逍遥", "6.5", "角色扮演"]
]
selected_row = small_table[1]
selected_score = selected_row[1]
print(selected_row)
print(selected_score)
print(small_table[1][1])
````

### 讲解与原始数据案例

第一步得到第二条记录`['一念逍遥', '6.5', '角色扮演']`，第二步得到评分文本'6.5'；连续索引得到相同值。

## 6. 列表切片：取出连续的一段

### 理论知识与语法

索引取一个元素，**切片（slice）取一段元素组成的列表**。本节使用连续切片：

```python
part = items[start:end]
```

规则是**包含start，不包含end**。省略start表示从开头开始，省略end表示到结尾。

| 写法 | 作用 |
| --- | --- |
| items[1:3] | 取索引1、2处的元素 |
| items[:2] | 取前两个元素 |
| items[1:] | 跳过第一个，保留其余元素 |

对列表切片会得到一个新的外层列表，原列表长度和顺序不变。本章只用它选取记录，不通过切片修改内层记录；不展开复制深度问题。

### 比较索引和切片的返回类型

````python
titles = ["人格解体", "一念逍遥", "明日方舟", "2048清", "炉石传说"]
print(titles[0], type(titles[0]))
print(titles[:1], type(titles[:1]))
print(titles[1:3])
````

### 讲解与数据案例

titles[0]得到一个名称字符串，titles[:1]得到只装了一个名称的列表；都“涉及第一个名称”，但返回对象不同。titles[1:3]取索引1、2，所以得到一念逍遥和明日方舟，不包括索引3。

sample已经是五条原始记录的简化表。切片既能截取名称列表，也能截取记录列表：

````python
first_two_records = sample[:2]
remaining_records = sample[1:]
print(first_two_records)
print("Remaining record count:", len(remaining_records))
print("Original record count:", len(sample))
````

### 易错点与练习

`sample[1:]`不会删除sample的第一行，而是返回另一个列表；要使用结果，需要接收它或直接处理表达式。切片越过右端会取到结尾，不像访问不存在的单个索引那样越界；例如短列表取[:5]不会自动补足五项。

## 7. open与with：文件对象及其使用范围

### 理论知识与语法

文件保存在磁盘上，列表存在于程序内存中。**open返回文件对象**，让程序能读取文件内容；文件对象不是已经读好的表格。

读取本章CSV采用：

```python
with open(path, "r", encoding="utf-8-sig", newline="") as file:
    # 在这里使用打开的文件
    print(type(file))
```

| 部分 | 含义 |
| --- | --- |
| path | 字符串形式的文件路径 |
| "r" | 以读取模式打开，不写入文件 |
| encoding="utf-8-sig" | 指定解码文字使用UTF-8 |
| newline="" | 让CSV读取工具按其规则处理换行 |
| as file | 将文件对象绑定给变量file |
| with后的缩进 | 文件保持打开、供本段代码使用的范围 |

with会在离开代码块时关闭文件，包括代码因异常退出该块的情况。file只是变量名，不是关键字。这里理解读取模式即可，不枚举其他文件模式。

若文件以 UTF-8 BOM 开头，用 encoding="utf-8" 读取时，表头第一列可能变成带不可见前缀的 id。本章统一使用 encoding="utf-8-sig"，读取时去掉 BOM。

**相对路径**以程序当前工作目录为起点，不保证总是以点击打开的文件位置为起点。本章约定从Notebook与CSV所在文件夹启动内核，这时文件名本身就是可用的相对路径。

### 代码

````python
with open("mobile_game_info.csv", "r", encoding="utf-8-sig", newline="") as file:
    print(type(file))
print("The with block has ended.")
````

## 8. csv.reader与list：解析记录和收集记录

### 理论知识与语法

**模块（module）提供可以使用的功能。** 本章用`import csv`加载标准库CSV工具，不需要安装额外的数据分析库。

```python
reader = csv.reader(file)
rows = list(reader)
```

这两行解决不同问题：

1. csv.reader读取CSV格式的记录，按分隔符、引号等规则解析字段；每次提供一条记录，记录是字符串组成的列表。
2. list(reader)把reader提供的记录收集起来，形成外层列表，才能方便地用rows[i][j]访问。

reader的读取位置会随着读取向后推进；同一个reader被完整收集后，不会自动回到文件开头。本章需要重读时重新打开文件，不把反复调用list(reader)当作重新读取。

CSV默认读取的字段是str，文件中的8.8不会自动变成float。CSV解析也不等于简单按每个逗号切割文本：被引号包住的字段可以含逗号，所以本章使用csv.reader而不是自己拆字符串。

### 代码：依次观察三种对象

## 10. for循环：语法、循环变量与执行顺序

### 理论知识

**循环让同一段代码重复执行。** for用于依次遍历可迭代对象中的元素。本章中列表和字符串都可以逐个提供元素，不需要先写出固定循环次数。

基本语法：

```python
for variable in iterable:
    statement_1
    statement_2
statement_after_loop
```

- for与in是语法组成部分；variable是当前元素使用的变量名。
- iterable是被遍历的对象；每轮提供一个元素。
- 冒号结束循环头，缩进语句组成循环体；本章统一四个空格。
- statement_after_loop没有缩进，等所有元素处理完成后才执行一次。

每轮的实际过程是：**取下一个元素 → 赋给循环变量 → 执行循环体**。下一轮把同一变量名绑定到另一个元素；它不会自动变成行号或序号。遍历空列表时，循环体一次也不执行。

### 代码：每轮处理一个名称

````python
titles = ["人格解体", "一念逍遥", "明日方舟"]
for title in titles:
    print(title)
print("Finished titles")
````

### 代码讲解

循环变量title先后得到三个名称。缩进的print执行三次，最后一句执行一次。代码的书写顺序没有变，但循环会重复执行同一个代码块。

遍历body时，每个元素是一整行，row便是一条七字段列表。名称和评分分别使用row[1]、row[2]。只需要一个循环：外层取记录，列索引取固定字段，不必再循环所有列。

下面直接使用已读入的完整文件，只显示其前五条，避免大量输出掩盖结构。

````python
for row in body[:5]:
    print(row[1], row[2])
print("Finished preview")
````

### 变量变化与易错点

| 轮次 | row代表 | row[1] | row[2] |
| --- | --- | --- | --- |
| 1 | 第一条七字段记录 | 人格解体 | '8.8' |
| 2 | 第二条七字段记录 | 一念逍遥 | '6.5' |
| 3 | 第三条七字段记录 | 明日方舟 | '8.2' |

row不是0、1、2。若把完成提示缩进，它会每轮重复；若把每行需要执行的print移到循环外，它只执行一次，通常只看到最后一次赋值的结果。

## 11. 类型转换与累加器：保存不断变化的总和

### 理论知识与语法

**累加器（accumulator）是用来保存逐步累计结果的变量。** 它是变量的一种使用方式，不是新的数据类型或Python关键字。

求和时需要两个步骤：

```python
total = 0.0             # 循环前建立起点
for value in values:
    total = total + value
```

右侧先读取旧total与当前value，计算结果，再重新赋给左侧total。也可以写`total += value`，但使用前total必须已初始化。本章先用完整写法看清过程。

如果value来自CSV，它通常是字符串，应先转换：`number = float(value)`。注意：

- 字符串加字符串是拼接，例如`"8.8" + "6.5"`得到文本`"8.86.5"`。
- 数值加字符串（如`0.0 + "8.8"`）会TypeError。
- float只能转换表示数值的值；对字段名`"score"`转换会ValueError，对一整行列表转换会TypeError。

### 代码：观察前三条评分的累计过程

````python
score_texts = ["8.8", "6.5", "8.2"]
running_total = 0.0
for text in score_texts:
    number = float(text)
    running_total = running_total + number
    print("Current score:", number, "Running total:", running_total)
````

### 代码讲解

总和依次为8.8、15.3、23.5。这说明每轮不仅使用当前值，也保留之前各轮的结果；浮点输出偶尔有末位尾差，不改变这些数学关系。

把同一规则用于文件记录，只需将“取出评分文本”明确写为row[2]。下例仍只对原文件前三条做演示：

````python
first_three_total = 0.0
for row in body[:3]:
    numeric_score = float(row[2])
    first_three_total = first_three_total + numeric_score
print("First-three total:", first_three_total)
````

### 易错点与练习

初始化只做一次，更新每轮做一次。如果把total=0.0放进循环体，之前的结果会不断被清空。重新运行整段统计代码时，也应重新执行初始化，避免延续上一次的总和。

## 12. 平均值：总和与分母必须对应同一批数据

### 理论知识与公式

**算术平均值**等于参与计算的所有数值之和，除以这些数值的个数：

```text
平均值 = 数值总和 / 数值个数
```

这里每条记录贡献一次评分。分母应数参与求和的记录，而不是表头、字段数量或某个随手写入的常数。循环结束后才拥有完整总和，随后计算平均值。

```python
average = total / len(selected_rows)
```

前提是selected_rows非空；空列表的长度为0，直接相除会ZeroDivisionError。本章的原文件和所选样本均非空。处理任意空结果需要后续的条件判断。

### 代码：同一批数值的总和与数量

````python
scores = [8.8, 6.5, 8.2]
three_score_total = 0.0
for score in scores:
    three_score_total = three_score_total + score
three_score_average = three_score_total / len(scores)
print("Mean of three scores:", three_score_average)
````

### 代码讲解与原文件案例

total累计了scores中的三项，len(scores)也只数这三项，因此总和与分母对应。把同样结构迁移到CSV：从同一个原文件读取全部记录，抽取前五条计算样本平均分。

这五条原始评分为8.8、6.5、8.2、9.9、7，数学总和40.4，平均值8.08。下列完整例子展示读取、选择、转换、累计、除法各自的位置。它计算的是**前五条样本**，不是全表结果。

````python
import csv

with open("mobile_game_info.csv", "r", encoding="utf-8-sig", newline="") as file:
    rows = list(csv.reader(file))
header = rows[0]
body = rows[1:]
selected_rows = body[:5]

sample_total = 0.0
for row in selected_rows:
    score = float(row[2])
    sample_total = sample_total + score
sample_average = sample_total / len(selected_rows)
print("Sample records:", len(selected_rows))
print("Sample total:", sample_total)
print("Sample mean:", sample_average)
````

### 易错点与练习

不能用前五条的总和除以完整文件的记录数，也不能用全表总和除以七个字段的数量。结果在评分范围内只能说明“看起来可能合理”，还需要检查列、选取范围、总和和分母。
