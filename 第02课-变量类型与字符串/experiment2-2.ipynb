## 任务二：按规则取名

&emsp;&emsp;在任务一中，我们了解到在一个项目中，在**语法规则内**可以为变量取任意我们想要的名称。甚至，在任务一的练习中，我们为我们的变量取名为 `asdf`。

&emsp;&emsp;但需要强调的是，我们的变量名称必须符合**语法规则**。例如，命名变量 `a data` 就报语法错误：

````python
a data = 50
````

&emsp;&emsp;`Python` 语法规定在变量名称中不允许使用空格字符。

&emsp;&emsp;在命名变量时，需要注意以下 `3` 个语法规则：

- **可以包括任何字母或数字（`a-z`，`A-Z`，`0-9`）以及下划线（`_`）**；
- **不能以数字开头**；
- **不能和其它 `Python` 保留的特殊名字冲突**。

&emsp;&emsp;其中，常见 `Python` 保留的特殊名字有：

- `for`, `while`, `break`, `pass`, `continue`
- `in`, `is`, `not`
- `if`, `else`, `elif`
- `def`, `class`, `return`, `yield`, `raises`
- `import`, `from`, `as`, `with`
- `try`, `except`, `finally`

&emsp;&emsp;另外，需要注意的是，变量名是区分大小写的，即 `data` 和 `Data` 是不同的。

&emsp;&emsp;我们举例看看以下变量名是否符合 `Python` 变量名的语法规则：

- `data`：合法
- `_data`：合法
- `Game_data`：合法
- `1data`：**不合法** (不能以数字开头)
- `I.am.data`：**不合法** (不能包含 `.`)
