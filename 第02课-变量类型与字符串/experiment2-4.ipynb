## 任务四：使用快捷方法实现加减乘除

&emsp;&emsp;在任务三中，我们尝试了 `data = data + 1`。实际上，在执行算术运算时，可以使用几种快捷方式来更新变量。

&emsp;&emsp;例如，对于 `data = data + 1`，我们可以编写成 `data += 1`：

```python
data = (2 + 7) * 4
print(data)
data += 1
print(data)
```

````python
# 练习区域

````

|   符号  | 原始写法 |             快捷写法                |
| ------------ | --- | ------------------------------- |
| += |  data = data + 1 | data += 1 |
| -= |  data = data - 1 | data -= 1  |
| *= |  data = data * 1 | data *= 1 |
| /= |  data = data / 1 | data /= 1|
| **= |  data = data ** 1 | data **= 1 |


&emsp;&emsp;注意，这些快捷操作符 `+=`，`-=`，`*=`，`/=`，`**=` 只能被用于变量的更新。这意味着这个变量在被更新前必须已经存储了一个值，即该变量必须已经被定义。

&emsp;&emsp;当我们对一个没有定义过的变量 `data1` 进行更新时，一定会报 `Python` 错误 `NameError`：

``` python
data1 += 1
```

&emsp;&emsp;输出报错：

``` text
---------------------------------------------------------------------------
NameError                                 Traceback (most recent call last)
<ipython-input-2-f38840d44420> in <module>
----> 1 data1 += 1

NameError: name 'data1' is not defined
```

````python
# 练习区域

````

&emsp;&emsp;注意，我们没有称这个错误为 “语法错误”，而是仅仅称其为 “错误”。这种错误与我们在第一个项目中了解到的语法错误不同，`data1 += 1` 是正确的 `Python` 语法，只是计算机返回错误，因为计算机无法更新尚未定义的 `Python` 变量。

&emsp;&emsp;只要语法正确，但计算机仍然出于某种原因而返回错误，就称为运行时错误。

&emsp;&emsp;另外一定要注意，我们更新变量的时候是这样书写的：`data += 1` 或 `data = data + 1`。在数学中，这两种写法都是错误的，因为 `data` 永远不会等于 `data + 1`。这也就意味着，计算机中的 `=` 运算符的含义与数学中的含义是不同的。在 `Python` 中，`=` 运算符是赋值的意思，即将右侧的值分配给左侧的变量，但并非是 “等于” 或 “平等” 的意思。

&emsp;&emsp;那什么时候会表示 “等于” 或 “平等” 的意思？可以使用 `==` 来表示平等的意思。举个简单的例子，`a == 10` 表示 `a` 等于 `10` 这个说法是否正确，如果正确返回 `True`，如果不正确则返回 `False`。当然，这里有包含好多你不懂的东西，我们后面会详细讲解，如果感兴趣，你现在可以在下方代码块中先试一试 `a == 10` 的输出(记得先对变量 `x` 进行赋值哦)：

``` python
a = 5
a == 10
```
