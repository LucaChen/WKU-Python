## 任务一：给数据取个名字

&emsp;&emsp;在第一个项目中，我们已经使用 `Python` 执行了一些算术运算。在本项目中，我们将对数值和文本数据进行保存。假设我们要保存计算机的计算结果供后续使用，那该怎么办？比如，我的身高是 `180 cm`，这个 `180 cm` 的数据在我们平时生活中就可以用 “我的身高” 这个名词来代替。

&emsp;&emsp;例如，`(2 + 7) * 4` 等于 `36`，我们想要保存 `36` 这个数字，是否可以用 “数据(data)” 来代表呢？在 `Python` 中，有变量和常量的概念，变量的意义在于帮助我们编写灵活易重复使用的代码，而不至于每次加减乘除的数值变了之后都需要重新编写一次代码。

&emsp;&emsp;那如何告诉计算机 `data` 就代表的是 `36`？

&emsp;&emsp;在 `Python` 中，符号 `=` 代表着 “赋值(`assignment`)” 的意思，变量在 `=` 左边，而用来赋值的对象在符号的右边：

``` python
data = 36# 为变量 `data` 赋值整数 36
```

````python
# 练习区域

````

&emsp;&emsp;我们可以看看是否赋值成功，我们打印 `data` 看看会输出什么：

```python
print(data)
```

````python
# 练习区域

````

&emsp;&emsp;计算机确实输出了我们刚刚赋值的数字 `36`。当计算机执行 `data = 36` 这行代码时，数字 `36` 将保存在计算机内存中。计算机内存具有许多存储位置，数字 `36` 被保存到某一个特定的位置。这个特定的存储位置具有唯一性，即可以像 `print(data)` 一样通过 `data` 这个变量访问数字 `36`。当然，`data` 作为一个变量名，可以取其他你想要的名字。

&emsp;&emsp;数字`36`被保存在一个特定的位置，想要看保存的位置需要引入一个新的概念`id`,通过`id(data)`来查看被赋值的数字36存储的位置:

```python
data = 36
print(id(data))
```

````python
# 练习区域

````

&emsp;&emsp;在第一个项目中，我们已经可以让计算机帮助我们计算 `(2 + 7) * 4`，我们将这个过程和赋值操作结合起来，让计算机在一个代码块中完成计算、赋值、输出的过程：

```python
data = (2 + 7) * 4
print(data)
```

````python
#练习区域
````

&emsp;&emsp;是的，计算机输出的是 `36` 而不是 `(2 + 7) * 4`。这是因为计算机首先计算 `(2 + 7) * 4`，然后将计算结果 `36` 保存到 `data`。

&emsp;&emsp;但是，好像这次验证不严谨。我们在之前就已经对 `data` 赋值了 `36`，那刚刚 输出的 `36` 真的是 `(2 + 7) * 4` 的计算结果吗？

&emsp;&emsp;答案是：是的。这个体验和验证交给大家！
