## 任务六：尝试把浮点数类型的数据转换成整数类型

&emsp;&emsp;既然 `Python` 中有 `int` 和 `float` 两种数字类型，那两种数字类型是否可以互转？

&emsp;&emsp;试想，如果一个数字是 `int` 整数类型，那么他只要在整数后面加上小数点和一个 `0`，应该就可以变成 `float` 类型。那么，如果已经是 `float` 数字类型了，要像变成 `int` 整数类型，好像必须去掉小数点及其之后的数字才能变成 `int` 整数类型，但是数值本身却依旧变化了。带着这两个问题，我们来尝试一下。

&emsp;&emsp;要将 `int` 整数转换为 `float` 浮点数，可以使用 `float()`，要将 `float` 浮点数转换为 `int` 整数，我们可以使用 `int()`：


&emsp;&emsp;看来确实是这样的，如果一个 `float` 浮点数要变成 `int` 整数类型，会将小数点及其之后的内容抹去。但是，这是直接抹去呢？还是四舍五入呢？直接尝试一下：


&emsp;&emsp;好吧，一个 `float` 浮点数使用 `int()` 的方法转换成 `int` 整数，是将小数点及其之后的内容直接抹去。但是如果想将 `float` 浮点数四舍五入为 `int` 整数类型，该怎么办？

&emsp;&emsp;让我们想想`float` 浮点数四舍五入的原理:小数部分小于`0.5`的数舍弃为`0`,大于等于`0.5`的数进`1`。一个 float 浮点数要变成 int 整数类型，会将小数点及其之后的内容抹去。如果我们将一个`float` 浮点数小数部分小于`0.5`的数再加`0.5`，该值的整数部分不会变，因此将其变成int 整数类型，会将小数点及其之后的内容抹去。如果我们将一个`float` 浮点数小数部分大于等于`0.5`的数再加`0.5`，该值的整数会进`1`，将其变成int 整数类型，小数点及其之后的内容也会被抹去。我们惊奇的发现，这与`float` 浮点数四舍五入的原理相似.

&emsp;&emsp;那将`float` 浮点数+`0.5`能否实现将 `float` 浮点数四舍五入为 `int` 整数类呢？

```python
int(1.2+0.5)
int(1.8+0.5)
```

&emsp;&emsp;我们可以发现上述方法可行。

&emsp;&emsp;还有其他方法来实现四舍五入么，答案是`round()`：
