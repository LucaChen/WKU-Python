## 任务八：给一句话取个名字

&emsp;&emsp;我们明白，字符串是以单引号'或双引号"括起来的任意文本。但有时候，我们需要创建的字符串中包含了带引号的字符串，那么字符串中的引号就会和字符串最外层的引号产生冲突。比如这句话：`Because that's just life is 'No pains,no gains'`

```python
string = 'Because that's just life is 'No pains,no gains''
print(string)
```

````python
# 练习区域

````

&emsp;&emsp;在这种情况下，将字符串放到单引号（`'`）内就会报语法错误 `SyntaxError: invalid syntax`。这个时候，我们就需要将双引号（`"`）替换单引号（`'`）：

```python
string = "Because that's just life is 'No pains,no gains'"
print(string)
```

````python
# 练习区域

````

&emsp;&emsp;仔细研究一下，我们用双引号将字符串包起来，这样看起来 `'No pains,no gains'` 就成为了字符串的一部分，这说明 `Python` 知道字符串在第二个双引号所在的位置结束。

![](Image/2_8_1.png)

&emsp;&emsp;我们提到了字符串内外都有单引号（'）的情况，那么字符串内外都有双引号（"）时又该怎么做？

```python
string = ""I love python""
```

&emsp;&emsp;在这种情况下，将字符串放到双引号（"）内就会报语法错误 SyntaxError: invalid syntax。这个时候，我们就需要将外层单引号（'）替换双引号（"）：

```python
string = '"I love python"'
print(string)
```

````python
# 练习区域
````

&emsp;&emsp;字符串中的引号就会和字符串最外层的引号产生冲突时可以改变字符串中的引号或字符串最外层的引号使字符串内外层引号不相同，就可避免冲突。

&emsp;&emsp;另外，`Python` 还有第二种方法完成这个字符串的赋值，在这句话中出现单引号（`'`）的地方之前输入一个反斜杠字符（`\`）：

```python
string = 'Because that\'s just life is \'No pains,no gains\''
print(string)
```

````python
# 练习区域

````

&emsp;&emsp;尝试到这里，我们是不是感觉现在思绪很混乱？

&emsp;&emsp;这里就要简述双引号和单引号的意义了，单引号（`'`）或双引号（`"`）本身只是一种表示方式，不是字符串的一部分。比如字符串 `'abc'` 只有 `a`，`b`，`c` 这 `3` 个字符。但是如果单引号或双引号为字符串的一部分，那么可以用转义字符 `\` 来标识。

&emsp;&emsp;转义字符 `\` 可以转义很多字符，比如 `\'` 表示 `'`，`\"` 表示 `"`，`\n` 表示换行，`\t` 表示制表符，如果字符 `\` 本身也要转义，所以 `\\` 表示的字符就是 `\`。

&emsp;&emsp;让我们来实际使用促进理解，你可以分别输入以下代码：

```python
str1 = 'I\'m \"Fine\"!'
print(str1)
str2 = 'I\'m a\nStudent.'
print(str2)
str3 = '\\\n\\'
print(str3)
```
