## 任务七：给单词取个名字

&emsp;&emsp;到目前为止，我们仅处理 `int` 和 `float` 值。但是，要想完成 `taptap` 和 `steam` 上数千个应用程序的数据分析项目，数字不是我们处理的唯一数据类型。例如，考虑下表，该表提供了有关来自 `mobile_game_info` 游戏的五个`id`的一些信息：

|   id   |  title   | score | game_type |   pulisher   | doenload | isAndroid |
| :----: | :------: | :---: | :-------: | :----------: | :------: | :-------: |
| 206369 | 人格解体 |  8.8  | 角色扮演  |  厂商：喵法  |  71069   |   TRUE    |

|  213   | 炉石传说 |   7   |   策略    | 厂商：Blizza | 10787305 |   TRUE    |

&emsp;&emsp;数据源为游戏数据集，已包含在每个项目目录下的 `assets/dataset` 中。我们发现，`title` 、`game_type`和`pulisher` 这三列使用的是不是纯数字表示。对应此类情况，在 `Python` 中我们可以将这些文本放到引号（" "）中完成我们的目标。现在，我们来定义游戏名称和游戏类型这两个变量：

```python
game_name = "人格解体"
game_type = "角色扮演"

print(game_name)
print(game_type)
```

&emsp;&emsp;`Python` 语法允许双引号（`"`）和单引号（`'`）。因此，如果我们要创建单词 `BlackWalnut`，则可以使用 "`BlackWalnut`" 或 '`BlackWalnut`'：

``` python
data1 = "BlackWalnut"
data2 = 'BlackWalnut'

print(data1)
print(data2)
```
