
&emsp;&emsp;没错，第二个输出的数据是小数。是的，在数学中，上述两个输出，一个是整数，一个是小数。同样，`Python` 也认可整数和小数的不同。

&emsp;&emsp;这里告诉大家，可以通过 `type()` 来查看数字类型，是整数还是小数：

``` python
data = (2 + 7) * 4
print(data)
print(type(data))
data = (56 + 86) / 54
print(data)
print(type(data))
```

````python
# 练习区域

````

&emsp;&emsp;的确，`36` 为 `int` 整数（`integer`）类型，而 `2.6296296296296298` 为 `float` 浮点数（`floating-point number`）类型。

&emsp;&emsp;`int` 和 `float` 虽然属于不同的数字类型，但在算术运算中将这些值混合在一起是不受限制的：

``` python
print(1+1.6)
print(4/2)
```
