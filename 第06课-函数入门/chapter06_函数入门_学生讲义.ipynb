## 1. `def` 是定义，不是执行

### 理论知识与语法

**函数是一包等到被叫到才运行的代码。** 基本形式：

```python
def function_name(parameter):
    statement
    return result
```

- `def` 这一行只是登记：以后可以用这个名字。
- 缩进的函数体在定义时**不执行**。
- `function_name(实际值)` 才是调用。

本课统一四个空格缩进。函数名不要用 `print`、`list`、`str` 这些内置名（第7课会专门做破坏性实验）。

### 案例：先定义平方，再调用


````python
def square(n):
    return n * n

print("After def, square has not printed 25 by itself")
print("square(5) =", square(5))
print("square(3) =", square(3))
````

### 讲解

定义 `square` 之后，若没有调用，不会出现 25。`square(5)` 把 5 传给 `n`，返回 25；`square(3)` 返回 9。同一段函数体可以服务不同输入。平方只为降低恐惧，真正要带走的项目函数是后面的抽列和频数。

### 易错点与练习

把 `def square(n):` 写成 `def square(5):` 是错的：参数位置写的是名字，不是某一次的实际数字。实际数字写在调用处。

## 3. `return` 不是 `print`

### 理论知识

**`return` 把值交回调用处，供后面继续使用。** 函数里的 `print` 只是把文字显示在屏幕上，并不把值交给 `=` 左边。

忘记 `return` 时，函数结束后默认交回 `None`。于是 `result = f(...)` 得到的是 `None`，再拿去累加或当列表用就会出问题。第8课清洗函数里最常见的误用，就是 `print` 了表却以为返回了表。

### 案例：对照“只打印”和“有返回”


````python
def square_print_only(n):
    print(n * n)

def square(n):
    return n * n

seen = square_print_only(5)
print("square_print_only returned:", seen)
value = square(5)
print("square returned:", value)
print("value + 1 =", value + 1)
````

### 讲解

`square_print_only(5)` 会在屏幕上看到 25，但 `seen` 是 `None`。`square(5)` 把 25 交到 `value`，所以 `value + 1` 可以计算。屏幕上有数字，不等于调用处拿到了数字。

### 易错点与练习

在函数末尾写 `print(result)` 却不 `return`，调用处无法把结果传给下一个函数。需要继续计算时必须 `return`。

## 5. 抽列函数 `extract_column`

### 理论知识

**抽列：给定表和下标，返回这一列所有格子组成的新列表。** 用 `for` 逐行 `append`，不要使用列表推导式。

```python
def extract_column(table, index):
    column = []
    for row in table:
        column.append(row[index])
    return column
```

换 `index` 就能换列：1 是名字，2 是评分，3 是类型。这就是本课验收要看的能力。

### 案例：抽出摘录类型列，再抽出全表前三个名字


````python
def extract_column(table, index):
    column = []
    for row in table:
        column.append(row[index])
    return column

sample_types = extract_column(sample, 2)
print("Sample types:", sample_types)
titles = extract_column(body, 1)
print(titles[0])
print(titles[1])
print(titles[2])
scores = extract_column(body, 2)
print("First three scores:", scores[0], scores[1], scores[2])
````

### 讲解

`extract_column(sample, 2)` 得到五个类型名，其中角色扮演出现在第 1、2、4 个位置。`extract_column(body, 1)` 的前三个名字应是人格解体、一念逍遥、明日方舟；评分列前三个是 8.8、6.5、8.2。不要一次打印整列几百个名字。

函数体里的 `table`、`index` 是参数名。真正的数据是调用处的 `sample` / `body` 和数字下标。

## 6. 把第5课四步收进 `freq_table`

### 理论知识

**不要重写频数逻辑，把四步缩进到 `def` 里。** 参数是“一列值”，返回字典。

四步仍是：空字典 → 循环 → 键不在则 0 → +1。本课不使用 `Counter`。

### 案例：对摘录类型列计数


````python
def freq_table(values):
    table = {}
    for value in values:
        if value not in table:
            table[value] = 0
        table[value] = table[value] + 1
    return table

sample_type_freq = freq_table(sample_types)
print(sample_type_freq)
print("角色扮演 in sample:", sample_type_freq["角色扮演"])
````

### 讲解

传入的是已经抽出的一列，不再在函数里写 `row[2]`。因此同一函数既能数类型，也能数平台。摘录里角色扮演 3 次。完整文件的类型频率表仍留给 P1，这里不要 dump 全表。

### 易错点与练习

`freq_table` 的参数是一维列表，不是整张二维表。若误传 `body`，键会变成整行列表，既不能当类型名，也难以验收。

## 8. 函数调函数：`column_freq`

### 理论知识

**一个函数的内部可以调用另一个已经写好的函数。** 频率表的常见组合是：先抽列，再频数。

```python
def column_freq(table, index):
    values = extract_column(table, index)
    return freq_table(values)
```

验收：同一对函数，只改 `index`，就能得到另一张频率表。不要复制两套四步。

### 案例：摘录类型频数，以及完整表的平台频数
