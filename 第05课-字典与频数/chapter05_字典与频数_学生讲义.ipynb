## 3. `in` 判断键，避免 `KeyError`

### 理论知识

**读取不存在的键会引发 `KeyError`。** 计数之前必须先问：这个键在不在字典里？

`key in d` 为 `True` 表示这个键已经有了；为 `False` 表示还没有。`not in` 是反过来的判断。

本课标准步伐是：若键不在，先把次数写成 `0`，再执行 `+1`。不要先 `+1` 再判断。

故障写法（只放在文字里，不要放进正常运行单元格）：

```python
# Deliberate fault: reading a missing key.
empty = {}
print(empty["角色扮演"])
```

### 案例：先问在不在，再取值或新建


````python
empty = {}
print("角色扮演" in empty)
print("角色扮演" not in empty)
if "角色扮演" not in empty:
    empty["角色扮演"] = 0
empty["角色扮演"] = empty["角色扮演"] + 1
print(empty)
print("策略" in empty)
````

### 讲解

空字典里还没有 `"角色扮演"`，所以 `in` 为 `False`，`not in` 为 `True`。先置 `0`，再 `+1`，字典变成 `{"角色扮演": 1}`。`"策略"` 仍不在，不要写 `empty["策略"]` 去打印次数。

上面的故障片段会 `KeyError`，因为读取时这个键还不存在。破坏性实验的意义是：先 +1 再判断，第一次见到新类型就会崩。

### 易错点与练习

`in` 检查的是**键**，不是值。`3 in freq_demo` 不是在问“有没有某个类型出现 3 次”。

1. **K3.1** 对 `freq_demo`，写出 `"策略" in freq_demo` 和 `"休闲" in freq_demo` 的结果。哪一个可以直接 `freq_demo[...]`，哪一个必须先置 0？

## 5. 频数四步：空字典 → 循环 → 置 0 → +1

### 理论知识与语法

**频数表就是“每个取值出现多少次”。** 本课把步伐写成四步板书，后面收成函数时也要原样保留：

1. 建空字典 `freq = {}`
2. `for` 循环每一行（或每一个值）
3. 若这个键不在字典里，先 `freq[key] = 0`
4. 无论新旧，执行 `freq[key] = freq[key] + 1`

不要为每个类型准备一个变量。不要使用 `Counter`，也不要使用字典推导式。本课要求你能亲手写出这四步。

### 案例：对五条摘录统计类型


````python
freq = {}
for row in sample:
    kind = row[2]
    if kind not in freq:
        freq[kind] = 0
    freq[kind] = freq[kind] + 1
print(freq)
````

### 讲解

循环五次。第一次见到 `"角色扮演"`，字典里没有它，先写成 0 再加 1，变成 1；第二次、第四次再遇到，直接 +1，最后是 3。`"策略"` 与 `"音乐"` 各出现一次。打印结果的 keys 是类型名。

完整七列表应写 `row[3]`。这里用摘录的 `row[2]`，因为简化表把类型放在下标 2。

### 易错点与练习

把 `if kind not in freq` 写反成 `if kind in freq: freq[kind] = 0` 会每次见到旧键都清零。缩进也要核对：置 0 只在键不存在时执行，+1 对每一行都执行。

## 7. 百分比：计数 / 总数 × 100

### 理论知识

**百分比 = 该键次数 / 全部次数 × 100。** 分子分母都是整数时，必须用 `/`（真正的除法）。`//` 是整数除法，`3 // 5` 得到 `0`，再乘 100 仍是 0，看起来像“这个类型一次都没有”。

总数应等于各键次数之和，也应等于参与统计的行数。百分比加起来应大约 100；浮点尾差可以接受，但 0 或 300 这种数通常说明除错了。

### 案例：把摘录频数换成百分比


````python
print("Integer division 3 // 5 =", 3 // 5)
print("True division 3 / 5 =", 3 / 5)
total = 0
for key in freq:
    total = total + freq[key]
print("Total from dict:", total)
print("Sample size:", len(sample))
percents = {}
for key in freq:
    percents[key] = freq[key] / total * 100
    print(key, freq[key], percents[key])
percent_sum = 0.0
for key in percents:
    percent_sum = percent_sum + percents[key]
print("Percent sum:", percent_sum)
````

### 讲解

`3 // 5` 是 0，`3 / 5` 是 0.6。角色扮演 3 / 5 × 100 = 60。策略与音乐各 20。三项合计 100。分母用字典次数之和，与 `len(sample)` 相同，因为摘录里每一行都进了频数表。

完整文件的百分比留给 P1。本格只对 5 条摘录演示除法口径。

### 易错点与练习

写 `freq[key] / total * 100` 时，若 `total` 仍是 0（循环一次都没加），会除零。先打印总数再算百分比。不要用 `int(百分比)` 提前截断后再求和。

## 8. 选做：最多、最少出现几次

### 理论知识

频数表做好之后，可以问“哪一类最多”。本课**不把 `min` / `max` 当作必会语法**；超时可以跳过。若使用，它们比较的是数字大小：`max(3, 1, 1)` 是 3。要找出对应的类型名，仍需要循环字典，比较每个键的次数。

下载量分箱（把下载量切成几档再计数）同样是选做，本课不展开。

### 案例：在摘录频数里找最多次数


````python
print("max of 3, 1, 1 =", max(3, 1, 1))
print("min of 3, 1, 1 =", min(3, 1, 1))
most_key = None
most_count = None
for key in freq:
    if most_count is None:
        most_key = key
        most_count = freq[key]
    else:
        if freq[key] > most_count:
            most_key = key
            most_count = freq[key]
print("Most common type in sample:", most_key, most_count)
````

### 讲解

摘录里最常见的类型是角色扮演，次数 3。`max(3, 1, 1)` 只告诉你最大数字，不告诉你名字。用循环才能同时记住键和值。
