## 任务四：注释某段代码使其不执行

&emsp;&emsp;经过之前的三个任务，如果你留意的话，你一定发现了计算机是从第一行代码开始向下执行的。并且，如果两行代码之间使用空格而不是换行，将报错。那么，如果两行代码之间使用空行而不是换行，又会发生什么，你可以以上一个任务的练习为例：

````python
# 练习区域

````

&emsp;&emsp;好像没有什么区别。确实，空行不会影响计算机执行代码。我们可以理解，虽然这个代码行空行了，但是这行代码行是存在的，只不过是空白，如果让计算机执行一行没有命令的代码，相当于没有输出，也就不会影响整体代码的执行了。

&emsp;&emsp;在 `Python` 中，除空白行外，计算机还忽略 `#` 符号右边的任何字符序列。在下方代码块中，我们可以看到，`print(23 + 7)` 之前被添加了 `#` 符号，执行代码的结果中不再输出结果 `30`。这是因为在 `Python` 中，`#` 符号后的内容称为代码注释，我们可以使用代码注释来添加我们对代码的备注。

```python
# print(23 + 7)
print(10 - 6)
print(12 + 38)
```

````python
# 练习区域

````

&emsp;&emsp;当然，代码注释还有其它写法，我们可以在代码行之后进行注释：

```python
# print(23 + 7)
print(10 - 6) # 将会输出 4
print(12 + 38)
```

````python
# 练习区域

````

&emsp;&emsp;在很多行都需要注释时，在每行都输入#来注释太麻烦，那么，有没有简便方法呢？选中需要注释的所有代码行，同时按下`Ctrl`键和`/`键，此时，所有被选中的代码行前都被添加上了`#`注释：

```python
# print(23 + 7)
# print(10 - 6) # 将会输出 4
# print(12 + 38)
```

````python
#练习区域
````

&emsp;&emsp;还有其他方法应对多行需要注释的情况么？将需要多行注释的区域用`'''`或`""" `包围:

```python

print(12 + 38)'''
```
