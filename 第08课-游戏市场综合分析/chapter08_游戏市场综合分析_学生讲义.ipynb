## 4. 去重：保留第一次出现，建新列表

### 理论知识

**遍历旧表，把第一次见到的标题追加到新表。** 用字典（或同样用 `not in` 的键表）记下已经见过的名字。不要修改正在 `for` 的那张列表。

```python
def dedupe_keep_first(body):
    seen = {}
    cleaned = []
    for row in body:
        title = row[1]
        if title not in seen:
            seen[title] = True
            cleaned.append(row)
    return cleaned
```

`seen[title] = True` 只表示“见过了”，真正留下的行在 `cleaned`。

### 案例：在很小的摘录上去重


````python
tiny = [
    ["一念逍遥", "6.5", "角色扮演", "1021860", "TRUE"],
    ["明日方舟", "8.2", "策略", "12263134", "TRUE"],
    ["一念逍遥", "6.5", "角色扮演", "999", "FALSE"],
]
seen = {}
cleaned_tiny = []
for row in tiny:
    title = row[0]
    if title not in seen:
        seen[title] = True
        cleaned_tiny.append(row)
print("tiny size:", len(tiny))
print("cleaned tiny size:", len(cleaned_tiny))
for row in cleaned_tiny:
    print(row[0], row[3], row[4])
````

### 讲解

三条变成两条：第一次出现的一念逍遥（TRUE，下载 1021860）留下，后面那条 FALSE 被丢掉。明日方舟只出现一次，直接留下。这张 `tiny` 的名字在下标 0，完整 `body` 的名字在下标 1。

全表去重后行数会下降，具体数字在 P1 用 `len` 打印。不要在作答栏填写教师提前给出的全表去重行数。

## 5. 不要在遍历时删除元素

### 理论知识

**正在 `for item in xs` 时，不要对 `xs` 做 `remove` / `pop` / `del`。** 解释器按位置往前走，你删掉当前元素后，下一格会错位，结果是漏删、漏处理，有时也不报错。正确做法：建新列表，把要留的 `append` 进去。

故障写法（只放文字，不要放进最终运行顺序）：

```python
# Deliberate fault: deleting while iterating.
names = ["一念逍遥", "明日方舟", "一念逍遥"]
for name in names:
    if name == "一念逍遥":
        names.remove(name)
```

### 案例：用新列表留下不等于目标的名字


````python
names = ["一念逍遥", "明日方舟", "一念逍遥"]
kept = []
for name in names:
    if name != "一念逍遥":
        kept.append(name)
print("original:", names)
print("kept:", kept)
````

### 讲解

原列表完整保留，便于对照。新列表只留下明日方舟。故障片段在真实运行中往往**不会**得到你以为的“两个一念逍遥都删掉”，这正是它危险的地方：看起来跑完了，结果却是错的。

去重、筛安卓都用同一纪律：旧表只读，新表只追加。

不要在讲义里公布全表“最热类型”的验收答案。P1 用清洗后的表自己算。

### 案例：摘录安卓子集上按类型加总下载量（演示用简化字段）


````python
# Simplified tiny android table: title, type, download
android_tiny = [
    ["人格解体", "角色扮演", "71069"],
    ["一念逍遥", "角色扮演", "1021860"],
    ["明日方舟", "策略", "12263134"],
]
type_freq = {}
download_sum = {}
for row in android_tiny:
    kind = row[1]
    amount = int(row[2])
    if kind not in type_freq:
        type_freq[kind] = 0
        download_sum[kind] = 0
    type_freq[kind] = type_freq[kind] + 1
    download_sum[kind] = download_sum[kind] + amount
print("type freq:", type_freq)
print("download sum:", download_sum)
````

## 8. 嵌套循环：外层类型，内层行

### 理论知识

**外层取出一个类型，内层再扫表，找这个类型的一条特征（例如第一个游戏名）。** 这是本课唯一略新的控制结构：两个 `for` 套在一起。不要使用 `break` 作为必会技巧；可以用“还没有找到例子”的标记，只记下第一次匹配。

### 案例：为摘录中每个类型找一个例子


````python
types = []
for row in sample:
    kind = row[2]
    if kind not in types:
        types.append(kind)
print("types in sample:", types)
for kind in types:
    example = ""
    for row in sample:
        if row[2] == kind:
            if example == "":
                example = row[0]
    print(kind, "example:", example)
````

### 讲解

先收集出现过的类型，避免依赖“类型列表已经写死”。内层对每个类型找出第一个名字：角色扮演 → 人格解体，策略 → 明日方舟，音乐 → 偶像梦幻祭2。全表应对清洗后的安卓表做同样的事，选几个类型预览即可，不必把每类全部游戏打出来。
