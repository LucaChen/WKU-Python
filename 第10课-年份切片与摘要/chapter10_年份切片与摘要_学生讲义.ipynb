## 1. 字符串和列表一样可以切片

### 理论知识

第 3 课对列表用过 `titles[:5]`。**字符串也可以按位置切开。** `"2000-11-01"[0]` 是第一个字符 `"2"`；`"2000-11-01"[:4]` 是前四个字符 `"2000"`。

发行日期这一列的形式是 `年-月-日`。本班统一用前四位当发行年，不要在课堂上每人一种 `split` 方案。切片下标含头不含尾：`[:4]` 取到下标 0、1、2、3。

### 案例：从日期文本取出年份


````python
date = "2000-11-01"
print(date[0])
print(date[:4])
print(date[:4] == "2000")
print("1999-04-01"[:4])
````

### 讲解

`"2000-11-01"[:4]` 等于 `"2000"`，比较结果为 `True`。`"1999-04-01"` 同理得到 `"1999"`。这就是后面 `row[2][:4]` 的迷你版：先取发行日期字段，再切前四位。

### 易错点与练习

## 3. 年份频率：还是那四步

### 理论知识

第 5 课四步：**空字典 → 循环 → 键不在则置 0 → +1**。第 6 课把它收进函数。本课可以重写四步，也可以自己再写一个 `freq_table`；不必从另一本笔记本 import。

键必须是四位年，例如 `"2000"`，而不是 `"2000-11-01"`，也不是 0、1、2。案例只统计前 5 行，避免把全部年份字典当成讲义答案。

### 案例：前 5 行的年份字典


````python
def freq_table(values):
    table = {}
    for value in values:
        if value not in table:
            table[value] = 0
        table[value] = table[value] + 1
    return table

years = []
for row in body[:5]:
    years.append(row[RELEASE_DATE][:4])
year_freq = freq_table(years)
print("First 5 years list:", years)
print("First 5 year freq:", year_freq)
for key in year_freq:
    print("Year key", key, "length", len(key), "count", year_freq[key])
````

### 讲解

前 5 行的键应是四位字符串：1999 出现 2 次，其余各 1 次。`len("1999")` 为 4，验收时看的就是这种键。全表年份频率在 P1 用全部 `body` 再做一遍；本格不要把全表字典贴进作答栏。

## 4. 不能对 int 切片，要先 `str()`

### 理论知识

切片是给**序列**用的：列表、字符串可以；整数不行。好评转换成 `int` 之后，若直接写 `n[0]`，会得到 `TypeError`。档位规则需要“看数字的第一位”，所以先 `str(n)`，再切片。

故障写法只留在文字里：

```python
# Deliberate fault: slicing an int.
n = 3318
print(n[0])
```

### 案例：先变成字符串再取首位


````python
n = 3318
text = str(n)
print(type(n), type(text))
print("First character:", text[0])
print("All characters:", text)
print("124534 as text[:1] / rest length:", str(124534)[0], len(str(124534)) - 1)
````

### 讲解

`3318` 的类型是 `int`，`str(3318)` 才是 `"3318"`。第一位 `"3"` 后面还有 3 位，档位里要补 3 个 0。`124534` 共 6 位，第一位 `"1"`，后面补 5 个 0。

## 5. 档位规则：首位、补 0、加号

### 理论知识

**档位是为了让频率表可读，不是为了精确算术。** 全班必须同一规则，否则两个人的字典无法对照：

1. 先把好评变成整数，再 `str(...)`。
2. 若只有 1 位，约定为 `"0+"`。
3. 否则：第一位 +（长度-1 个 `"0"`）+ `"+"。

因此 `3318` → `3000+`，`124534` → `100000+`。不要改成四舍五入或自己发明的区间。

### 案例：两个规定例子


````python
def rating_bucket(n):
    text = str(int(n))
    if len(text) == 1:
        return "0+"
    return text[0] + "0" * (len(text) - 1) + "+"

print("3318 ->", rating_bucket(3318))
print("124534 ->", rating_bucket(124534))
print("40 ->", rating_bucket(40))
print("5 ->", rating_bucket(5))
````

### 讲解

`3318` 变成 `3000+`，`124534` 变成 `100000+`。`40` 是两位，变成 `40+`；个位数走 `"0+"` 分支。P1 的档位频率字典里应能看到 `100000+` 这类键——具体次数你自己统计，讲义不预填全表。

### 易错点与练习

补 0 的个数是 `len(text) - 1`，不是固定 3 个。位数不同的数补错，是本课卡点。
